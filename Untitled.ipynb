```python
import pandas as pd 
import numpy as np 
from faker import Faker
import random 
from datetime import datetime, timedelta
```

```python
# Initialize faker (using Indian locale for realistic name/places if needed)
fake=Faker('en_In')
Faker.seed(42)
random.seed(42)
np.random.seed(42)


# Defining our Consonants 
NUM_ORDERS = 10000
STORES=['Gurgaon_Sec43','Noida_Sec63', 'Delhi_Vasantkunj', 'Dwarka_Sec12', 'South_Ex']
CATEGORIES=['Grocery', 'Fresh Vegetables', 'Electronics', 'Snacks', 'Personal Care']
```

#### Generating *Orders* table

```python
orders_data = []
start_date = datetime(2026, 9, 1) # Simulating data for Sep-Oct 2026
end_date = datetime(2026, 10, 30)

for i in range(NUM_ORDERS):
    order_id = f"ORD{100000 + i}"
    customer_id = f"CUST{random.randint(1000, 5000)}"
    store_id = random.choice(STORES)

    # generating random timestamp within our date range 
    order_time = fake.date_time_between(start_date=start_date, end_date=end_date)

    total_amount = round(random.uniform(99.0, 1500.0), 2)

    #  95% Delivered, 5% Cancelled
    status=np.random.choice(['Delivered', 'Cancelled'], p=[0.95, 0.05])

    orders_data.append({
        'Order_ID': order_id,
        'Customer_ID': customer_id,
        'Store_ID': store_id,
        'Order_Timestamp': order_time,
        'Total_Amount': total_amount,
        'Order_Status': status
        
    })

df_orders = pd.DataFrame(orders_data)
print("Orders table created!")

        
```

#### Generating *Deliveries* Table

```python
deliveries_data = []

# Only creating delivery records for orders that were actually Delivered
delivered_orders = [order for order in orders_data if order['Order_Status'] == 'Delivered']

for i, order in enumerate(delivered_orders):
    delivery_id = f"DEL{200000 + i}"
    order_id = order['Order_ID']
    rider_id = f"RIDER{random.randint(1, 200)}"
    order_time = order['Order_Timestamp']
    
    distance_km = round(random.uniform(0.5, 4.5), 1)
    
    # --- ANOMALY INJECTION LOGIC ---
    # If the order is placed between 7 PM (19:00) and 9 PM (21:00), simulate operational delays
    if 19 <= order_time.hour <= 21:
        pick_delay = random.randint(4, 8)       # Slower packing due to high volume
        transit_time = random.randint(8, 15)    # Slower transit due to traffic
    else:
        pick_delay = random.randint(1, 3)       # Normal 1-3 mins packing
        transit_time = random.randint(4, 7)     # Normal 4-7 mins transit
    
    # Add a slight penalty for longer distances
    if distance_km > 3.0:
        transit_time += 2
        
    picked_time = order_time + timedelta(minutes=pick_delay)
    dispatched_time = picked_time + timedelta(minutes=random.randint(1, 2)) # Handoff to rider
    delivered_time = dispatched_time + timedelta(minutes=transit_time)
    
    deliveries_data.append({
        'Delivery_ID': delivery_id,
        'Order_ID': order_id,
        'Rider_ID': rider_id,
        'Picked_Timestamp': picked_time,
        'Dispatched_Timestamp': dispatched_time,
        'Delivered_Timestamp': delivered_time,
        'Distance_km': distance_km
    })

df_deliveries = pd.DataFrame(deliveries_data)
print("Deliveries table created with anomalies!")
```

### Generate *Inventory* and *Order_Defects* Tables

```python
# 5a. Create Inventory Data
inventory_data = []
for store in STORES:
    for sku_num in range(1, 51): # 50 products per store
        sku_id = f"SKU{str(sku_num).zfill(3)}"
        category = random.choice(CATEGORIES)
        stock_qty = random.randint(10, 500)
        velocity = random.randint(1, 50) # Daily sales speed
        
        inventory_data.append({
            'SKU_ID': sku_id,
            'Store_ID': store,
            'Category': category,
            'Stock_Quantity': stock_qty,
            'Daily_Sales_Velocity': velocity
        })
df_inventory = pd.DataFrame(inventory_data)

# 5b. Create Order Defects Data (Only 4% of delivered orders have defects)
defects_data = []
defect_types = ['Missing Item', 'Damaged Packaging', 'Wrong Item Delivered', 'Quality Issue']

num_defects = int(len(delivered_orders) * 0.04)
defective_orders = random.sample(delivered_orders, num_defects)

for i, order in enumerate(defective_orders):
    defects_data.append({
        'Defect_ID': f"DEF{300000 + i}",
        'Order_ID': order['Order_ID'],
        'Defect_Type': random.choice(defect_types),
        'Refund_Amount': round(order['Total_Amount'] * random.uniform(0.1, 1.0), 2) # Refund between 10% to 100%
    })
df_defects = pd.DataFrame(defects_data)
print("Inventory and Defects tables created!")
```

### Exporting to *CSV* for *SQL*

```python
# Saving all tables to CSV in your current working directory
df_orders.to_csv('Orders.csv', index=False)
df_deliveries.to_csv('Deliveries.csv', index=False)
df_inventory.to_csv('Inventory.csv', index=False)
df_defects.to_csv('Order_Defects.csv', index=False)

print("All datasets successfully exported! Ready for SQL ingestion.")
```
